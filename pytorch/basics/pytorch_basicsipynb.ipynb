## Pytorch tensor

- just like numpy has arrays, pytorch has tensors that we work with.

```python
# 3x3
W = [[1, 1, 1],
     [1.5, 1.5, 1.5],
     [2, 2, 2]]

# 3x1
x = [[6], [7], [8]]
# 3x1
b = [[1], [1], [1]]

# Variable to store output
# 3x1
y = [[0], [0], [0]]
```

```python
# In pytorch, we can convert these lists of lists into tensors
# we can define the tensor as float as the initilization phase in two different ways
W_torch = torch.FloatTensor(W)

# when we define the tensor with 'torch.tensor' and did not determine the type of tensor,
# it automatically assigns a type of tensor based on the list of lists (float,long,..)
x_torch = torch.tensor(x).float()

# Lets use the function "ones" to create an array of ones!
b_torch = torch.ones(3, 1)

# Lets now compute Wx + b using these numpy variables!
output = torch.matmul(W_torch, x_torch) + b_torch

# Print out the result
print("Output:\n", output)
print("Output shape:\n", output.shape)
```

# Pytorch basic operations

## Element wise operations

```python
# creating a 2D tensor using torch.rand
y = torch.rand(4,5)
print("Our 2D tensor is: \n", y)

# performing python scalar arithmetic on tensors
print("\nScalar Multiplication:\n",y * 10)
print("Addition and Square:\n",(y + 1)**2)
print("Addition:\n",y + y)
print("Addition and division:\n",y / (y + 1))

# We can use a combination of Torch functions and normal python arithmetic
print("\nPower and square root:\n",torch.sqrt(y**2))

# torch tensors are objects and they have their own functions
print(f"Y -\n min: {y.min():.2f}\n Max: {y.max():.2f}\n std: {y.std():.2f}\n sum: {y.sum():.2f}")
```

## Tensor operations

- Tensor operations are a bit different to that of array operations. especially multiplications.

- Tensor multiplications are perfromed as **batch operations**. (like a stack).

- This help improve the parallalization of the multiplicaton operations (useful in gpus).

- When we have two arrays of 3D matrices, numpy multiplies them at a time but what id= fwe have 10 arrays. Pytorch stacks then 10 tensors and multiplies them parallely as batches.


```python
# Create two 3D Tensors
tensor_1 = torch.rand(3,3,3)
tensor_2 = torch.rand(3,3,3)

# Add the 2 Tensors
print("Addition:\n",tensor_1 + tensor_2)

# We cannot perform a normal "matrix" multiplication on a 3D tensor
# But we can treat the 3D tensor as a "batch" (like a stack) of 2D tensors
# And perform normal matrix multiplication independently on each pair of 2D matrices (3- 2D matrices in each tensor)
print("Batch Multiplication:\n", torch.bmm(tensor_1, tensor_2))
```

```python
# Lets create a more interesting tensor: A 3D tensor
tensor_3 = torch.rand(2,4,5)

# We can swap the Tensor dimensions
print("\nThe origional Tensor is is:\n", tensor_3)
print("With shape:\n", tensor_3.shape)

# Transpose will swap the dimensions it is given
# Here the 0,2 are the indeces of the dimensions itself
# (meaning, the dim0 and dim2 are swapped).making it from (2,4,5) to (5,4,2)
print("The Re-arranged is:\n", tensor_3.transpose(0,2))
print("With shape:\n", tensor_3.transpose(0,2).shape)
```

## Indexing
- Indexing is almost same in Numpy

```python
# Create a 4D Tensor
tensor = torch.rand(2,3,1,4)
print("Our Tensor:\n",tensor)

# Select the last element of dim0
'''
The '-1' in this case denotes the last element in the dim0 (2).
- we have a, 2 * (3,1,4) matrices. when we specify '-1',
we get the last (3,1,4) matrix (tensor) as output
'''
print("\nThe last element of dim0:\n",tensor[-1])

# 1st element of dim0 [0]
# 2nd element of dim1 [1]
print("\nIndexed elements:\n",tensor[0, 1])

# Select all elements of dim0 [:]
# The 2nd element of dim1 [1]
# The 1st element of dim2 [0]
# The 3rd element of dim3 [2]
print("\nIndexed elements:\n",tensor[:, 1, 0, 2])
```

## Describing our tensors
- Same as numpy

```python
# Lets create a large 4D Tensor
tensor = torch.rand(3, 5, 3, 2)

# View the Number of elements in every dimension
print("The Tensor's shape is:", tensor.shape)

# In Pytorch shape and size() do the same thing!
print("The Tensor's shape using size() is:", tensor.size())

# View the number of elements in total
print("There are %d elements in total:" % tensor.numel())

# View the number of Dimensions
print(f"There are {tensor.ndim} Dimensions")
```

## Reshaping

- reshape the tensor to different shape keeping the number of elements same.

- same as numpy. and '-1' is the place holder to tell torch that that dimension is flexibe to adapt the number of elements to fit the required shape of the output tensor

```python
# Let us reshape our Tensor to a 2D Tensor
print("Reshape to 3x30:\n", tensor.reshape(3, 30))

# We can also use the Flatten method to convert to a 1D Tensor
print("Flatten to a 1D Tensor:\n",tensor.flatten())

# Here the -1 tells Pytorch to put as many elements as it needs here in order to maintain the given dimension sizes
# AKA "I don't care the size of this dimension as long as the first one is 10"
print("Reshape to 10xwhatever:\n",tensor.reshape(10, -1))
```

## Squeezing and unsqueezing

- To increase the number of dimensions of a tensor (maybe to fit the shape of an input to a model (batch processing)), we commonly add a **'empty dimension'**, to the tensor.

- In the same way we can also remove an empty dimension from the tensor.

```python
# Lets create a 2D Tensor
tensor = torch.rand(3, 2)

# View the Number of elements in every dimension
print("The Tensors shape is:", tensor.shape)

# Unsqueeze adds an "empty" dimension to our Tensor
'''As in the case of reshaping, the '2' here denotes the dimesion where
we want to unsqueeze (adding an empty dimension)
since  our tensor is a 2D, 2 here indicates that we want to unsqueeze
at the end of the tensor making it as 3D tensor
(3,2) -->(3,2,1)
'''
print("Add an empty dimenson to dim3:", tensor.unsqueeze(2).shape)

# Unsqueeze adds an "empty" dimension to our Tensor at dim1 (3,2)-->(3,1,2)
print("Add an empty dimenson to dim2:", tensor.unsqueeze(1).shape)
```

```python
# Lets create a 4D Tensor with a few "empty" dimensions
tensor = torch.rand(1, 3, 1, 2)

# view the Number of elements in every dimension
print("The Tensors shape is:", tensor.shape)

# squeeze removes an "empty" dimension from our Tensor (1,3,1,2) -> (1,3,2)
print("Remove empty dimenson dim3:", tensor.squeeze(2).shape)

# squeeze removes an "empty" dimension from our Tensor
print("Remove empty dimenson dim0:", tensor.squeeze(0).shape)

# If we don't specify a dimension, squeeze will remove ALL empty dimensions
# (1,3,1,2) -> (3,2) Therefore, be careful when using squeeze without specifying dimension
print("Remove all empty dimensons:", tensor.squeeze().shape)
```
