# Why LSTM's and GRU's

- The appealing feature of RNN's is its ability to connect the information from previous steps to current output and so on.. However, RNN's fail to connect information across long sequences (Inputs)
- The main problem with RNN's is that when the input sequence is longer the gradients of the starting inputs gets less and less untill where it takes a long time to adjust the weights of these layers!
- This problem is called as vanishing gradients. Normally we back propogate our error in NN

- To address this problem many variations of RNN's have been proposed in the literature.
- Some of the famous ones are LSTM's and GRU's

## LSTM and GRU!
- It stands from Long Short Term Memory. It is a special king of RNN that is capable of learning long term dependencies.
- It introduces 3 gates or valves to remember, update and output the dependencies over long sequences.
- [Christopher Olah's](https://colah.github.io/posts/2015-08-Understanding-LSTMs/) blog is a very good and intutive to learn more about LSTM's.
- LSTM hav three gates namely:
  1. Forget Gate: Decides which info to forget
  2. Input Gate: Decides which info to add
  3. Output Gate: Decides which info to output.

![LSTM](https://colah.github.io/posts/2015-08-Understanding-LSTMs/img/LSTM3-chain.png#center)

- In the figurewe can see the three modules, those are from three timesteps $(t-1, t, t+1)$.
![gates](../assets/Images/LSTM.JPG)
- Terms:
  1. $C_{t-1}$: is where our memory flows from previous to current timestep and so on.., it is called as memory cell.
  2. $h_{t-1}$: is the previous hidden state
  3. $C_t$: the memory cell state at the current next time step
  4. $h_t$: The current hidden state and the output (we feed this to the next step and also to the world!)
  5. $x_t$: current input.
  - All these terms are Vectors and the sigmoid and tanh is where the NN operations take place(weights Update!)
> - An important note is that there is no Neural Network operatrions on the memory cell, meaning there wont be any gradients to vanish over long sequences.
- All the Neural Network ooperations happen on the cells below. i.e on the hidden state and input side.
- The first sigmoid layer is the Forget Gate and it is just: $f_t = σ(W_f ⋅[h_{t-1}, x_t] + b_f)$
  - It looks at the $h_{t-1}$ and $x_t$ and outputs a number b/w 0 and 1 for every element in the $C_{t-1}$ vector.
- The second gate is the 'Input layer'. This decides what new information that is allowed to store in the memrory state. First the sigmoid layer decides which values we'll add next to the memory cell, then tanh layer creates a vector of new candidate values $\tilde{C}_t$ that could be added to the state. Then these two will be combined to create an update to the state.
  - $i_t = σ(W_i \cdot [h_{t-1}, x_t] + b_i)$
  - $\tilde{C}_t = tanh(W_c \cdot [h_{t-1}, x_t] + b_c)$
- So we with these layers we can update our memory cell from previous state $C_{t-1}$ to current state $C_t$
  - We multiply the old state with $f_t$ and forget the thing we decide earlier and then we add $i_t * \tilde{C}_t $ (The new candidates vales scaled by importance of each value): $ \bbox[yellow]{C_t} =  f_t * C_{t-1} + i_t * \tilde{C}_t$
- Finally we get to the output gate where we output the values in filtered version.
  - First the current state is passes through a **tanh** to make the values b/w -1 and 1, the we also multiply it with the output of the sigmoid layer
  - It is given by $o_t = σ(W_o [h_{t-1}, x_t] + b_o)$ and $\bbox[yellow]{h_t} = o_t * tanh(C_t)$

- GRU's aka, Gated Recurrent Neural Networks are a little fancy versions of LSTM's. They combines the forget gate and input gates into a single gate called 'update gate'.
- Another modification is it merges the cell state and hidden state which results in a simpler model than LSTM!!
